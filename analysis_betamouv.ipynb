```python
n_trial = 200
n=0

for trial in tqdm(project.trialgroup.trials, desc="cropping"):
    if n>n_trial: 
        break
    if not trial.is_valid() or trial.date.month == 5 or any(stat["is_valid"] == False for bp, stat in trial.trajectories.items()):
        continue

    save_as: Path = output_dir / f"{trial.name}.csv"
    # if save_as.exists(): 
    #     continue

    raw_tracking = DLCTracking(path=trial.pred_path,
                                view=trial.camera_view, 
                                bodyparts=BODYPARTS,
                                time_pad_off=trial.time_pad_off,
                                cm_per_pixel=trial.cm_per_pixel,
                                shift=trial.camera_shift)

    clean_tracking = DLCTracking(path=trial.clean_pred_path,
                                view=trial.camera_view, 
                                bodyparts=BODYPARTS,
                                time_pad_off=trial.time_pad_off,
                                cm_per_pixel=trial.cm_per_pixel,
                                shift=trial.camera_shift, 
                                status=trial.trajectories)

    raw_coords_set = raw_tracking.load(False)
    clean_coords_set = clean_tracking.load()

    for bp in BODYPARTS: 
        if bp not in clean_coords_set.keys(): 
            clean_coords_set[bp] = raw_coords_set[bp]
            clean_coords_set[bp].coords = clean_coords_set[bp].crop_xy(start=trial.time_pad_off - 0.1, end=trial.time_pad_off + 0.4)
            continue

        traj = clean_coords_set[bp] 
        traj.coords = traj.crop_xy(start=trial.time_pad_off - 0.1, end=trial.time_pad_off + 0.4)
        
        # since it in cm, put back in pixel, without removing the shift
        traj.coords["x"] = traj.coords["x"] / trial.cm_per_pixel
        traj.coords["y"] = traj.coords["y"] / trial.cm_per_pixel

    # format saving
    data = {bp: traj.coords.set_index("t")
        for bp, traj in clean_coords_set.items()}

    feat_df = pd.concat({raw_tracking.scorer: pd.concat(data, axis=1)}, axis=1, names=["scorer", "bodyparts", "coords"])
    if  len(feat_df) != 63: 
        continue

    # Replace index with frame_num (0, 1, ..., len(feat_df)-1)
    feat_df.index = pd.RangeIndex(start=0, stop=len(raw_df), name="frame_num")

    raw_df.to_csv(output_dir / f"{trial.name}.csv")
    trial.save_yaml(output_dir / f"{trial.name}.yaml")
    n+=1

```

```python
import numpy as np
import plotly.express as px
import plotly.graph_objects as go


def plot_tracking(feat_df, skeleton, indices=None, delta_T=10, by_label=False, color_by_time=True) :
    bodyparts = bodyparts_to_use

    if indices is None:
        sampled = feat_df.iloc[::delta_T]
        title = f"Bodypart tracking (every {delta_T} frames)"
    else:
        sampled = feat_df.loc[indices] if by_label else feat_df.iloc[indices]
        title = f"Bodypart tracking ({len(sampled)} selected frames)"
    
    n = len(sampled)
    t_norm = np.linspace(0.1, 1.0, n)  # 0.3 -> skip the near-white end of the scale
    colors = px.colors.qualitative.Plotly

    fig = go.Figure()

    # --- Skeleton ---
    if color_by_time:
        # One trace per sampled frame, colored along a sequential scale
        n = len(sampled)
        frame_colors = px.colors.sample_colorscale(
            "Blues", [i / max(n - 1, 1) for i in range(n)])

        for k, (idx, row) in enumerate(sampled.iterrows()):
            xs, ys = [], []
            for a, b in skeleton:
                xs += [row[a]["x"], row[b]["x"], None]
                ys += [row[a]["y"], row[b]["y"], None]
            fig.add_trace(go.Scatter(
                x=xs, y=ys, mode="lines",
                line=dict(width=1.5, color=frame_colors),
                opacity=0.8,
                hoverinfo="skip",
                name=f"t={idx}",
                legendgroup="skeleton",
                showlegend=False,
            ))

    else:
        for a, b in skeleton:
            seg_colors = px.colors.sample_colorscale(color_scale_map[a], t_norm)

            xa, xb = sampled[a]["x"].to_numpy(), sampled[b]["x"].to_numpy()
            ya, yb = sampled[a]["y"].to_numpy(), sampled[b]["y"].to_numpy()

            for k in range(n):
                fig.add_trace(go.Scatter(
                    x=[xa[k], xb[k]],
                    y=[ya[k], yb[k]],
                    mode="lines",
                    line=dict(width=5, color=seg_colors[k]),
                    opacity=1,
                    hoverinfo="skip",
                    showlegend=False,
                ))

    # --- Points (drawn after, so they sit on top) ---
    for i, bp in enumerate(bodyparts):
        point_colors = px.colors.sample_colorscale(color_scale_map[bp], t_norm)
        fig.add_trace(go.Scatter(
            x=sampled[bp]["x"], y=sampled[bp]["y"],
            mode="markers",
            name=bp,
            marker=dict(size=10, color=point_colors, opacity=0.9,
                        line=dict(width=0.5, color="black")),
        ))

    fig.update_layout(
        title=title,
        xaxis_title="X", yaxis_title="Y",
        template="plotly_white", width=900, height=800,
    )
    fig.update_yaxes(
        # autorange="reversed", 
                     scaleanchor="x")
    fig.write_html(u.make_path(output_dir / "figs", f"{path.stem}" ))
    # fig.savefig(u.make_path(output_dir / "figs", f"{path.stem}.png" ))
    
    return fig
```

#### PCA at 90% explained

```python

pca90 = PCA(n_components=0.90)
X_pca = pca90.fit_transform(X_scaled)

n_cp = X_pca.shape[1]
col = [f"PC{i}" for i in range(1, n_cp+1)]

# 3. Store results in a new DataFrame
pca_df = pd.DataFrame(
    X_pca,
    columns=col,
    index=label_df.index
).join(label_df)

pca_df.head(3)
```

```python
pca90.n_components_
```

```python
eigenvalues = pca90.explained_variance_
explained_var = pca90.explained_variance_ratio_ * 100
cumulative_var = np.cumsum(explained_var)

n_components = len(eigenvalues)
components = np.arange(1, n_components + 1)

fig, ax1 = plt.subplots(figsize=(10, 6))

# Eigenvalues (bars)
ax1.bar(
    components,
    eigenvalues,
    color="steelblue",
    alpha=0.7,
)

ax1.set_xlabel("Principal component")
ax1.set_ylabel("Eigenvalue", color="steelblue")
ax1.tick_params(axis="y", labelcolor="steelblue")
ax1.set_xticks(components)

# Cumulative explained variance
ax2 = ax1.twinx()

ax2.plot(
    components,
    cumulative_var,
    color="darkorange",
    marker="o",
    linewidth=2,
)

ax2.set_ylabel("Cumulative explained variance (%)", color="darkorange")
ax2.tick_params(axis="y", labelcolor="darkorange")
ax2.set_ylim(0, 105)

# elbow
ax1.axhline(1, color="black", linestyle="--", alpha=0.5)
ax1.text(6, 1.1, "elbow", color="black", alpha=0.5, size=12)

ax2.axhline(90, color="darkred", linestyle="--", alpha=0.5)
ax2.text(6, 91, "90% threshold", color="darkred", alpha=0.5, size=12)

ax1.set_title("Scree plot")
plt.tight_layout()
```

```python
def HMM(data: pd.DataFrame, feature_cols: list[str], n_state: int) -> pd.DataFrame:
    """Run HMM on data (must only contain qualitative data) with n_state
    Returns the same data dataframe, with the column 'pred_state' containing the n states 
    predicted by the model""" 

    from hmmlearn.hmm import GaussianHMM
    from sklearn.preprocessing import StandardScaler

    X = StandardScaler().fit_transform(data[feature_cols].to_numpy())
    lengths = data.groupby("name", sort=True).size().to_numpy()   # frames per trial
    assert lengths.sum() == len(X)

    # several random starts, keep the best (EM finds local optima)
    best, best_score = None, -np.inf
    for seed in range(5):
        m = GaussianHMM(n_components=n_state, covariance_type="diag",
                        n_iter=200, random_state=seed)
        m.fit(X, lengths)
        score = m.score(X, lengths)

        print(f"    seed: {seed}   score: {score}")
        if score > best_score:
            best, best_score = m, score
    model = best

    data["pred_state"] = model.predict(X, lengths)
    print(np.round(model.transmat_, 2))

    return data
```

```python
import numpy as np
import plotly.express as px
import plotly.graph_objects as go


def plot_tracking(feat_df, skeleton, indices=None, delta_T=10, by_label=False, color_by_time=True) :
    bodyparts = bodyparts_to_use

    if indices is None:
        sampled = feat_df.iloc[::delta_T]
        title = f"Bodypart tracking (every {delta_T} frames)"
    else:
        sampled = feat_df.loc[indices] if by_label else feat_df.iloc[indices]
        title = f"Bodypart tracking ({len(sampled)} selected frames)"
    
    n = len(sampled)
    t_norm = np.linspace(0.1, 1.0, n)  # 0.3 -> skip the near-white end of the scale
    colors = px.colors.qualitative.Plotly

    fig = go.Figure()

    # --- Skeleton ---
    if color_by_time:
        # One trace per sampled frame, colored along a sequential scale
        n = len(sampled)
        frame_colors = px.colors.sample_colorscale(
            "Blues", [i / max(n - 1, 1) for i in range(n)])

        for k, (idx, row) in enumerate(sampled.iterrows()):
            xs, ys = [], []
            for a, b in skeleton:
                xs += [row[a]["x"], row[b]["x"], None]
                ys += [row[a]["y"], row[b]["y"], None]
            fig.add_trace(go.Scatter(
                x=xs, y=ys, mode="lines",
                line=dict(width=1.5, color=frame_colors),
                opacity=0.8,
                hoverinfo="skip",
                name=f"t={idx}",
                legendgroup="skeleton",
                showlegend=False,
            ))

    else:
        for a, b in skeleton:
            seg_colors = px.colors.sample_colorscale(color_scale_map[a], t_norm)

            xa, xb = sampled[a]["x"].to_numpy(), sampled[b]["x"].to_numpy()
            ya, yb = sampled[a]["y"].to_numpy(), sampled[b]["y"].to_numpy()

            for k in range(n):
                fig.add_trace(go.Scatter(
                    x=[xa[k], xb[k]],
                    y=[ya[k], yb[k]],
                    mode="lines",
                    line=dict(width=5, color=seg_colors[k]),
                    opacity=1,
                    hoverinfo="skip",
                    showlegend=False,
                ))

    # --- Points (drawn after, so they sit on top) ---
    for i, bp in enumerate(bodyparts):
        point_colors = px.colors.sample_colorscale(color_scale_map[bp], t_norm)
        fig.add_trace(go.Scatter(
            x=sampled[bp]["x"], y=sampled[bp]["y"],
            mode="markers",
            name=bp,
            marker=dict(size=10, color=point_colors, opacity=0.9,
                        line=dict(width=0.5, color="black")),
        ))

    fig.update_layout(
        title=title,
        xaxis_title="X", yaxis_title="Y",
        template="plotly_white", width=900, height=800,
    )
    fig.update_yaxes(autorange="reversed", scaleanchor="x")
    fig.write_html(u.make_path(analysis_dir / "single_trial" / f"{path.stem}" , title), auto_open=True)
    
    return fig
```
